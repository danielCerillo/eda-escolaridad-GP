```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
file_path = "Niveles-de-escolaridad-de-la-poblacion-de-15-anos-y-mas-en-Gomez-Palacio--Distribucion-de-la-poblacion-total.csv"
df = pd.read_csv(file_path)
df
```

```python
df.info()
```

```python
print(df.isnull())
```

```python

print(df.describe())
```

```python
df["Percentage"] = df["Percentage"].round(2)
df_sorted = df.sort_values(by="Population", ascending=False)
```

```python
plt.figure(figsize=(10, 6))
top_5 = df_sorted.head(5)
plt.bar(top_5["Academic Degree"], top_5["Population"], color="steelblue")
plt.title("Top 5 Academic Degrees in Gómez Palacio (15+ years old)")
plt.xlabel("Academic Degree")
plt.ylabel("Population")
plt.xticks(rotation=45, ha='right') # Rotate labels to avoid text overlapping
plt.tight_layout()
plt.show()
```

```python
higher_ed_filter = ["Licenciatura", "Especialidad", "Maestría", "Doctorado"]
higher_ed = df[df["Academic Degree"].isin(higher_ed_filter)]

plt.figure(figsize=(8, 5))
plt.bar(higher_ed["Academic Degree"], higher_ed["Population"], color="darkred")
plt.title("Higher Education Distribution in Gómez Palacio")
plt.xlabel("Degree Level")
plt.ylabel("Population")
plt.show()
```
